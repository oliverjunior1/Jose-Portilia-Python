```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
%matplotlib inline
```

```python
np.random.seed(14)

x = list(range(1,6))
y = np.linspace(1,10,5)

fig, ax = plt.subplots()

def makeBars(ax, x, y, w):
    for x, y in zip(x, y):
        bar = Rectangle((x,0),w, y)
        ax.add_patch(bar)

makeBars(ax, x, y, 0.4)

ax.autoscale()  
```

```python
fig.set_size_inches(15,5)

fig
```

```python
ax.set_title('Some bar char')
ax.set_xlabel('This i9s x label')
ax.set_ylabel('Thids is a y label')

fig
```

```python
help(Rectangle)
```

```python
for r in ax.patches:
    r.set_edgecolor('g')
    r.set_ls('dashed')
    r.set_lw(3)
    r.set_hatch('\\')

fig
```

```python
sketches = [0,0.25,0.5,0.75,1]

for r, s in zip(ax.patches, sketches):
    r.set_sketch_params(s)

fig
```

```python
rec_id = range(10)

zorder =reversed(rec_id)

for r, z in zip(rec_id, zorder):
    ax.patches[r].set_zorder(z)

fig
```
